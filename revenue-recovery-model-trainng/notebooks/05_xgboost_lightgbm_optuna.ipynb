# C6 — XGBoost + LightGBM + Optuna

**Primary tuning metric: PR-AUC.**

Why not accuracy? The positive class is only about 22–23%, so raw accuracy can
be dominated by the majority class.

Model-selection order:
1. Cross-validated PR-AUC (case-grouped and stratified)
2. Validation PR-AUC
3. ROC-AUC as a secondary ranking metric
4. Log Loss and Brier Score for probability quality

The test set is not used for tuning/model selection.

````python
# Execute the reproducible C6 experiment from this notebook's project root.
# Or run: python scripts/run_boosting_optuna.py

from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "data" / "c4_splits").exists():
    ROOT = ROOT.parent

print("Project root:", ROOT)
print("C4 train:", pd.read_csv(ROOT / "data/c4_splits/train.csv").shape)
print("C4 validation:", pd.read_csv(ROOT / "data/c4_splits/validation.csv").shape)
print("C4 test:", pd.read_csv(ROOT / "data/c4_splits/test.csv").shape)
````
